Cleanup

```python
import re

i = 1
with open (path, "r") as f_read, open (f"{data_path}/quotes_cleaned.txt", "w") as f_write:
    for quote in f_read:
        quote = quote.strip(" \n")
        quote = re.sub(" +", " ", quote)
        print (quote)

        if quote != "":
            f_write.write(f"{i}. {quote}\n")
            i += 1
            



```
