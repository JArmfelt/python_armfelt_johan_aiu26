# while_statments

- loops when condition is True
- "if-loop" kör sålänge det är true

```python
n = 0
while n < 10:
    print(f"Hello {n}")
    n += 1  
```

```python
i = 0

while i <= 5:
    print(f"hi number {i}")
    i += 1
```

first 10 sqares

```python
i = -10
while i <= 10:
    # **2 är upphöjt till 2, alltså kvadraten av i
    square = i**2
    print(square, end="  ")
    i += 1
```

## Oil leakage

```python
year = 0
birds = 80000

while birds > 8000:
    birds = birds / 2
    year += 1
    print(f"Year {year}: {birds} birds left") 
```
