## Iterate trough list with for

```python
for tv_show in tv_shows:
    print(tv_show)
```

```python
for i, tv_show in enumerate(tv_shows,1):
    print(i, tv_show)
```

# List comprehension

```python
cubes = [x**3 for x in range(10)]
cubes
```

```python
[x**2 for x in range(10) if x**2 > 25]
```

Plot graph

```python
import matplotlib.pyplot as plt

x = list(range(10))
x
```

```python
k = 2
m = 2
y = [k*x +m for x in x]
y
```

```python
plt.plot (x,y, 'o--')
plt.title("Plot y = 2x+2")
```
