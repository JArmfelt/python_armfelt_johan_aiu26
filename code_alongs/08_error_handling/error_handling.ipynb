try - except

```python
age = input("Enter age: ")

try:
    age = float(age)
    if not 0 < age < 120:
        raise ValueError("Age must be between 0 and 120")
        print(f"Age is: {age} years")
except ValueError as e:
    print(f"Invalid input: {e}")
    
```

```python
while True:
    age = input("Enter age: ")
    try:
        age = float(age)
        if not 0 < age < 120:
            raise ValueError("Age must be between 0 and 120")
        print(f"Age is: {age} years")
        break
    except ValueError as e:
        print(f"Invalid input: {e}")
```
